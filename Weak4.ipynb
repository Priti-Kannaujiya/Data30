```python
for col in df.columns:
    if df[col].dtype != 'object':  # only numeric
        skew_val = df[col].skew()
        if abs(skew_val) > 1:
            df[col] = df[col].replace(0, df[col].median())
        else:
            df[col] = df[col].replace(0, df[col].mean())
```

```python
zero_count = (df == 0).sum()
print(zero_count)
```

```python
plt.figure(figsize=(15, 8))
df[num_cols].boxplot()
plt.title("Boxplot Before Scaling")
plt.xticks(rotation=45)
plt.show()
```

```python
scaler = StandardScaler()               # performing standard scaling
df_scaled = df.copy()
df_scaled[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
plt.figure(figsize=(15, 8))
df_scaled[num_cols].boxplot()
plt.title("Boxplot After Scaling (Standardized Data)")
plt.xticks(rotation=45)
plt.show()
```

```python
col = 'trestbps'  # we can change this to any numeric column
Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1
outliers_iqr = df[(df[col] < (Q1 - 1.5 * IQR)) | (df[col] > (Q3 + 1.5 * IQR))]

# Z-score method
z_scores = np.abs((df[col] - df[col].mean()) / df[col].std())
outliers_z = df[z_scores > 3]

print(f"\nOutliers using IQR for '{col}':", outliers_iqr.shape[0])
print(f"Outliers using Z-score for '{col}':", outliers_z.shape[0])
```
