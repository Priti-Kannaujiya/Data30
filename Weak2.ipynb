## Day 10

```python
import matplotlib.pyplot as plt
```

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Sales': [1200, 1500, 1700, 1600, 1900, 2100, 2500, 2400, 2600, 3000, 3200, 3500]
}

df=pd.DataFrame(data)
df.head()
```

### Ploting a line chart

```python
plt.figure(figsize=(10, 6))
plt.plot(df['Month'], df['Sales'], color='blue', marker='o', linestyle='-', linewidth=2, markersize=6)
```

### Customize titles and axis

```python
plt.figure(figsize=(10, 6))
plt.plot(df['Month'], df['Sales'], color='blue', marker='o', linestyle='-', linewidth=2, markersize=6)
plt.title("Monthly Sales Growth in 2025", fontsize=16, fontweight='bold', color='darkblue')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales (in USD)", fontsize=12)
```

## Day 12

```python
data=pd.read_csv("iris.data.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data=data.drop_duplicates()
data.duplicated().sum()
```

```python
data.describe()
```

### Hist plot

```python
data.hist(figsize=(10, 8), color='skyblue', edgecolor='black')
plt.suptitle("Feature Distributions in Iris Dataset", fontsize=14)
plt.show()
```

### Box plot

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=data, palette='Set2')
plt.title('Boxplots of Iris Features')
plt.show()
```

### Pair plot

```python
sns.pairplot(data, hue='Iris-setosa', palette='Set1')
plt.suptitle("Pairplot of Iris Features by Species", y=1.02, fontsize=14)
plt.show()
```

### Heat Map

```python
plt.figure(figsize=(6, 4))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Iris Features')
plt.show()
```

## Day 13

```python
import numpy as np
from scipy import stats

x = np.random.randn(50)  # sample 1
y = np.random.randn(50) + 1  # sample 2 (shifted mean)

# Mean comparison (Independent t-test)
t_stat, p_val = stats.ttest_ind(x, y)
print("T-test p-value:", p_val)

# Correlation test (Pearson)
corr, p_corr = stats.pearsonr(x, y)
print("Correlation:", corr)
```
